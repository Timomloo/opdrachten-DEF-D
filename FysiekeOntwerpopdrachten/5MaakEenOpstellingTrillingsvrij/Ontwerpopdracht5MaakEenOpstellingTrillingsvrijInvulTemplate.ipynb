## Opdracht 2b: Opsplitsen taak 2: afleiding ontwerpeis

```python
# Bereken hier je benodigde kantelfrequentie

f = 5 #Hz
f_0 = 0.5 * f #na berekening
print('De kantelfrequentie is',f_0,'Hz.')
```

```python
# Voeg hier je plot in.

omega_0 = 2 * np.pi * f_0

def func(freq):
    return np.abs(1 / (1 - ((2*np.pi*freq)/omega_0)**2))

x_test = np.linspace(1,150,1000)

plt.figure()
plt.plot(x_test,func(x_test),'-',label = 'Verwachte Grafiek')
plt.xlabel('Frequentie (Hz)')
plt.ylabel('|$A_m / A_v$|')
plt.xscale('log')
plt.yscale('log')
plt.legend()
plt.show()
```

```python
# Bereken hier de minimale Delta-x waarbij je aan de eis van de opdracht voldoet.
g = 9.81

dx_min = g / (4*np.pi**2 * f_0**2)

print('De mininmale delta x waarde is gelijk aan %.5f m.' %dx_min)
```

## Opdracht 3: Plot de grafiek.

```python
# Voeg hier je plot in. Lees goed de opdracht wat er in de grafiek moet staan!
# Het is ook handig om je grafiek op te slaan, doe dat door met plt.savefig().

massa = np.linspace(0,1,1000)

def veerconstante(m,dx):
    return m / dx * g

plt.figure()
plt.plot(massa,veerconstante(massa,dx_min),label='Berekende deltaX')
plt.plot(massa,veerconstante(massa,2*dx_min),label='deltaX maal 2')
plt.plot(massa,veerconstante(massa,5*dx_min),label='deltaX maal 5')
plt.xlabel('Massa (kg)')
plt.ylabel('Veerconstante (N/m)')
plt.legend()
plt.grid()
plt.savefig('Opdracht3plot.png')
plt.show()
```
